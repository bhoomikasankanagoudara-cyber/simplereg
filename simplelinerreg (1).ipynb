```python
#### vertically dividing dataset as input and o/p
X=df.iloc[:,0]
Y=df.iloc[:,1]
```

SPLITING DATASET AS TRAINING AND TESTING TEST

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,x_test =train_test_split(X,Y,test_size=0.2,random_state=0)
```

```python
len(x_train)
```

```python
x_train
```

```python
import numpy as np
np.array(x_train).reshape(-1,1)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
### training the model
lr=LinearRegression()
lr.fit(x_train.values.reshape(-1,1),y_train)
```

```python
x_test.values
```

```python
### predicting
lr.predict(x_test.values.reshape(-1,1))
```

```python
y_test.values
```

```python
####  evaluation model

from sklearn.metrics import r2_score, mean_absolute_error,root_mean_squared_error,r2_score
root_mean_squared_error(y_train,lr.predict(x_train.values.reshape(-1,1)))
```

```python
r2_score(y_train,lr.predict(x_train.values.reshape(-1,1)))
```
